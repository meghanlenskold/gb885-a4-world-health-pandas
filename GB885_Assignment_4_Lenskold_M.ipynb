Scenario: You are an analyst for a global infant and maternal health advocacy group called JPG. You task is to use your understanding of Pandas to provide Josephina with the information she needs.

```python
# Import libraries.
import numpy as np
import pandas as pd
```

```python
# Create dataframes.
world_health_url = "https://raw.githubusercontent.com/katiegaertner/ML-Fundamentals/main/world-health-data-numbers.csv"
world_health_df = pd.read_csv(world_health_url)
```

```python
# Question 1: How many rows are in the DataFrame you created?
# Question 2: How many columns are in the DataFrame you created?

world_health_df.shape
```

```python
# Preview first 5 columns.
world_health_df.head()
```

```python
# Set index.
world_health_df = world_health_df.set_index("Abbreviation")

world_health_df = world_health_df.sort_index()

world_health_df.head()
```

```python
# Question 3: What nation has the index value "DZ"?

world_health_df.loc["DZ"]
```

```python
# Drop Vatican City from our records.

world_health_df = world_health_df.drop("VA")
```

```python
# Question 4: What is the official language of the nation on the 186th row?

world_health_df.iloc[185]["Official Language"]
```

```python
# Question 5: View the 10 nations with the highest infant mortality--what continent holds these ten nations?

world_health_df = world_health_df.sort_values("Infant mortality_per_1k", ascending=False)

world_health_df.head(10)
```

```python
# Question 6: Create a new dataframe, but only for countries whose official language is Spanish. How many rows are in your new dataframe?

spanish_speaking_df = world_health_df[world_health_df["Official Language"] == "Spanish"]

spanish_speaking_df.shape
```

```python
# Question 7: What is the mean infant mortality per 1,000 infants in countries whose official language is Spanish?

print(spanish_speaking_df["Infant mortality_per_1k"].mean())
```

```python
# Question 8: What is the median maternal mortality in countries whose official language is Spanish?

print(spanish_speaking_df["Maternal_Mortality_per_100k"].median())
```

```python
# Question 9: Create a new column for GDP per capita, and identify the nation with fourth-highest GDP per capita.

world_health_df["GDP_Per_Capita"] = world_health_df["GDP"] / world_health_df["Population"]

world_health_df.sort_values("GDP_Per_Capita", ascending=False).head(5)
```

```python
# Question 10: What is the maternal mortality rate per 100,000 live births for that nation?

print(world_health_df.loc["CH", "Maternal_Mortality_per_100k"])
```

```python
# Question 11: What is the the average maternal mortality rate per 100,000 for all nations in the data set?

print(world_health_df["Maternal_Mortality_per_100k"].mean())
```

```python
# Question 12: What percentage of health care is paid out of pocket by country with the lowest infant mortality?

world_health_df.sort_values("Infant mortality_per_1k", ascending=True).head(1)
```

```python
# Question 13: What percentage of health care is paid out of pocket by country with the highest infant mortality?

world_health_df.sort_values("Infant mortality_per_1k", ascending=False).head(1)
```
